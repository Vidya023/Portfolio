```python
import pandas as pd 
import numpy as np 
import yfinance as yf 
```

```python
data = pd.read_html('https://en.wikipedia.org/wiki/Nasdaq-100')[4]
data.head() 
```

```python
tickers = data['Ticker'].tolist() 
```

```python
df = yf.download(tickers,start='2010-01-01')['Adj Close']
```

```python
df.head() 
```

```python
df.dropna(axis=1,inplace=True) 
```

```python
df.head(3) 
```

Calculating monthly return

```python
ret_mtl = (df.pct_change()+1)[1:].resample('ME').prod() 
ret_mtl.head(5)
```

Defining function to get 3 month, 6 month and 12 month returns

```python
def get_rolling_ret(df, n):
    return df.rolling(n).apply(np.prod) 
```

```python
ret_12 = get_rolling_ret(ret_mtl,12)
ret_12.dropna(inplace=True) 
ret_12.head(3) 
```

```python
ret_6 = get_rolling_ret(ret_mtl,6)
ret_6.dropna(inplace=True) 
ret_6.head(3) 
```

```python
ret_3 = get_rolling_ret(ret_mtl,3)
ret_3.dropna(inplace=True) 
ret_3.head(3) 
```
